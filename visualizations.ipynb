# Visualizations

In order to provide a better understanding of the data, there is the possibility to create different types of visualizations. The following visualizations are available:
- Gaze overlay
- Time Series plot
- Heatmap
- Main sequence plot

## Prerequisites
In order to use this notebook, you need a fully preprocessed dataset. If you haven't preprocessed your data yet, please preprocess the dataset first. Refer to the documentation on how to process datasets: https://multipleye-cost.github.io/multipleye-preprocessing/getting_started/#getting-started

It is also possible to use a dataset that someone else has preprocessed for you. You can specify the path of the preprocessed data in one of the cells below. The preprocessed data should contain a `multipleye_settings_preprocessing.yaml` file, which contains the settings for the dataset.

## Loading a preprocessed dataset

```python
from pathlib import Path
from preprocessing import data_collection
from preprocessing import settings
import preprocessing
from preprocessing.plotting.plots import create_plots
```

```python
data_collection_path = Path("/Users/debor/repos/multipleye-preprocessing/preprocessed_data/MultiplEYE_PL_PL_Warsaw_1_2025")
```

```python
# load the settings for this dataset. for preprocessed datasets th config is always in the same place, not need to change the path
settings.load_from_yaml(data_collection_path / "multipleye_settings_preprocessing.yaml")
```

```python
dc = data_collection.MultipleyeDataCollection.create_from_preprocessed(Path(data_collection_path))
```

## Load data for one session

In this notebook you can look at the plots for individual sessions to show the functionality. You can either just pick a random session or pick one based on the session identifier. Execute only the respective cell.

```python
# pick a session based on the session identifier
identifier = "007_PL_PL_1_ET1" # e.g., 007_PL_PL_1_ET1
sess = dc[identifier]
```

```python
# pick a random session. you can just adapt the index on all sessions to look at another one
all_sessions = list(dc)
sess = all_sessions[0]
```

```python
gaze = preprocessing.load_trial_level_raw_data(
                sess.sid,
                trial_columns=settings.TRIAL_COLS,
                load_metadata=True,
            )
```

```python
gaze = preprocessing.load_trial_level_events_data(
    gaze,
    sess.sid,
    event_type=settings.FIXATION,
    file_pattern=None,
)

gaze = preprocessing.load_trial_level_events_data(
    gaze,
    sess.sid,
    event_type=settings.SACCADE,
    file_pattern=None,
)
```

# Main sequence plot

The main sequence plot visualizes the relationship between saccade amplitude and peak velocity. The plot shows the saccades as points, where the x-axis represents the amplitude of the saccade and the y-axis represents the peak velocity. The plot can help to identify any abnormalities in the saccadic eye movements. The distribution of this relationship should always follow a specific pattern. Please consult additional literature on this topic for more information, such as Holmqvist et al. (2011) "Eye Tracking: A Comprehensive Guide to Methods and Measures" or Bahill et al. (1975) "Main sequence: a tool for studying human eye movements".

```python
create_plots(
    gaze,
    "main_sequence",
    session_identifier=sess.session_identifier,
    directory=Path(settings.OUTPUT_DIR) / "visualizations",
    show=False
)
```

# Gaze overlay

The gaze overlay plot allows for visualizing the gaze samples and fixations on the stimulus image. The raw gaze samples are shown as a line while the fixations are visualized as dots, the size of the dots corresponds to the duration of the fixation.

```python
create_plots(
    gaze,
    "gaze_overlay",
    stimuli=sess.stimuli,
    session_identifier=sess.session_identifier,
    directory=Path(settings.OUTPUT_DIR) / "visualizations",
    show=True,
)
```
